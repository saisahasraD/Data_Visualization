```python
import requests

url = "http://quotes.toscrape.com"
response = requests.get(url)

print(response.status_code)
html = response.text
print(html)
```

This code uses the Python requests library to access a webpage. It sends a GET request, checks the response status code, and extracts the webpage's HTML content using response.text.v something like this

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, "html.parser")
quotes = soup.find_all("div", class_="quote")

print(f"Found {len(quotes)} quotes on this page")
```

This code uses BeautifulSoup to parse the webpage's HTML, finds all quote elements (<div> tags with class "quote"), stores them in a list, and prints the total number of quotes found.

```python
data = []
for q in quotes:
    text = q.find("span", class_="text").get_text()
    author = q.find("small", class_="author").get_text()
    tags = [t.get_text() for t in q.find_all("a", class_="tag")]
    data.append({"quote": text, "author": author, "tags": tags})

print(data)
```

This code loops through all the quotes, extracts the quote text, author name, and tags, stores them as dictionaries in a list called data, and prints the extracted information.

```python
import pandas as pd

df = pd.DataFrame(data)
df.head()

df.to_csv("quotes.csv", index=False)
```

This code converts the extracted data into a Pandas DataFrame, displays the first five records, and saves the data as a CSV file named quotes.csv.

```python
import matplotlib.pyplot as plt

df["author"].value_counts().plot(kind="bar", color="#028090")
plt.title("Quotes per Author")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup


url = "https://www.aceec.ac.in/academics/computer-science-and-engineering-aiml-2/"
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')

tables = soup.find_all('table')
print(f"Found {len(tables)} table(s) on the page.")


if tables:

    dfs = pd.read_html(str(tables))

    for index, df in enumerate(dfs):
        print(f"\n--- Table {index + 1} ---")
        print(df.head())


else:
    print("No HTML <table> elements were found on this specific URL. (The layout might use CSS grids or lists instead).")
```
